```python
import pandas as pd

data = {
    "Hours": [1, 2, 3, 4, 5, 6, 7, 8, 9],
    "Marks": [20, 25, 35, 40, 50, 60, 65, 80, 90]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.linear_model import LinearRegression

X = df[["Hours"]]
y = df["Marks"]

model = LinearRegression()
model.fit(X, y)

print("Model trained successfully!")
```

```python
hours = 5

prediction = model.predict(
    pd.DataFrame({"Hours": [hours]})
)

print("Study Hours:", hours)
print("Predicted Marks:", round(prediction[0], 2))
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["Hours"], df["Marks"], label="Actual Marks")
plt.plot(df["Hours"], model.predict(X), label="Prediction Line")

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Student Marks Prediction Using ML")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error

predictions = model.predict(X)

mae = mean_absolute_error(y, predictions)

print("Mean Absolute Error:", round(mae, 2))
```
